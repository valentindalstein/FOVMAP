```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib qt

from fovmap.file_utils import parse_lens_dir_csv_file
from fovmap.file_utils import rotate_vectors
from fovmap import FILES_PATH
```

```python
filename = FILES_PATH / 'ommatidia_lens_dir.csv'
right_eye = parse_lens_dir_csv_file(filename)
right_eye_rotated = rotate_vectors(right_eye, 0, 90, 0)
```

```python
right_eye_color = 'red'
rotated_right_eye_color = 'blue'
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
# Create a Poly3DCollection from the faces
ax.scatter(*zip(*right_eye.tolist()), color=right_eye_color, s=1)
ax.scatter(*zip(*right_eye_rotated.tolist()), color=rotated_right_eye_color, s=1)
# Set labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_box_aspect([1, 1, 1])
ax.set_title('Eye and rotated eye')
ax.legend(['Right eye', 'Rotated right eye'])
plt.show()
```
