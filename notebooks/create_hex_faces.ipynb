```python
from copy import deepcopy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import json
import os
import datetime
%matplotlib qt

from fovmap.file_utils import parse_lens_dir_csv_file
from fovmap.manual_hex_segmentation import manual_hex_segmentation
from fovmap import FILES_PATH
```

Read file

```python
filename = FILES_PATH / 'ommatidia_lens_dir.csv'
vertices = parse_lens_dir_csv_file(filename)
vertices = vertices.tolist()
initialized = False
```

Initialize

```python
if not initialized:
    current_cp = vertices[0]
    cps_vertices = []
    hgps_vertices = []
    edge_vertices = []
    vertices_wo_cps_edges = deepcopy(vertices)
    vertices_wo_cps_hgps_edges = deepcopy(vertices)
    cp_sel_angle_threshold = 10
    hgps_sel_angle_threshold = 7.5
    n_center = 0
    ommatidia_hex_grid = {}
    initialized = True
```

Compute

```python
(
    current_cp,
    cps_vertices,
    hgps_vertices,
    edge_vertices,
    vertices_wo_cps_edges,
    vertices_wo_cps_hgps_edges,
    cp_sel_angle_threshold,
    hgps_sel_angle_threshold,
    n_center,
    ommatidia_hex_grid
) = manual_hex_segmentation(
        current_cp,
        cps_vertices,
        hgps_vertices,
        edge_vertices,
        vertices_wo_cps_edges,
        vertices_wo_cps_hgps_edges,
        cp_sel_angle_threshold,
        hgps_sel_angle_threshold,
        n_center,
        ommatidia_hex_grid
        )
```

Save

```python
# Save the ommatidia_hex_grid
current_date = datetime.datetime.now().strftime('%Y-%m-%d_%H-%M-%S')

file_path = FILES_PATH / f'{current_date}_{filename.stem}_mhs.json'
if os.path.exists(file_path):
    overwrite_check = input(f'{file_path} already exists. Overwrite? [y/n]')
    if overwrite_check == "y":
        with open(file_path, 'w') as f:
            json.dump(ommatidia_hex_grid, f, indent=4)
else:
    with open(file_path, 'w') as f:
        json.dump(ommatidia_hex_grid, f, indent=4) 
```
