```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib qt
from fovmap.surface_plot import cone_surface_apex_neg_z, cone_surface_apex_neg_z_w_blckd_ngl
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
# Plot the cone
x,y,z = cone_surface_apex_neg_z(
radius=50,
cone_angle_deg=48.791,
angle_revolution_deg=340,
angle_offset_deg=10,
height=58.796,
height_cutoff=33,
z_base_offset=0,
resolution=100
)
ax.plot_surface(x, y, z, color='b', alpha=0.5)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_box_aspect([1, 1, 1])
ax.set_xlim([-50, 50])
ax.set_ylim([-50, 50])
ax.set_zlim([-50, 50])
# Set title
ax.set_title('Plotting cone')
# Set the view angle
ax.view_init(elev=30, azim=90, roll=0)
plt.show(block=True)
```
